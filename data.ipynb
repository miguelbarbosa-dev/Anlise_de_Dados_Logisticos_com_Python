```python
df = df.dropna()
display(df)
```

```python
display(df[["preco","id_produto"]].head(306))
display(df.info())
#Formatos errados, todos com excessão do ID_Produto estão em STR
```

```python
#Padronização e formatando os tipos de dados corretamente
df["validade"] = pd.to_datetime(df["validade"], format="mixed")
df["validade"] = pd.to_datetime(df["validade"], format="%Y-%m-%d")
df["preco"] = pd.to_numeric(df["preco"].astype("string").str.replace("R$","",regex=False).str.replace(" ","",regex=False).str.replace(",", ".",regex=False).str.replace("preçopendente", "",regex=False))
df["estoque_unidades"] = pd.to_numeric(df["estoque_unidades"].astype("string").str.replace(" ", "",regex=False).str.replace("un","",regex=False))



print(df.info())
display(df.head(100))
```
