```python
df = df.dropna()
display(df)
```

```python
display(df[["preco","id_produto"]].head(306))
display(df.info())
#Formatos errados, todos com excessão do ID_Produto estão em STR
```

```python
#Padronização e formatando os tipos de dados corretamente
df["validade"] = pd.to_datetime(df["validade"], format="mixed")
df["validade"] = pd.to_datetime(df["validade"], format="%Y-%m-%d")
df["preco"] = pd.to_numeric(df["preco"].astype("string").str.replace("R$", "",regex=False).str.replace(" ", "",regex=False).str.replace(",", ".").str.replace("preçopendente", "",regex=False))
df["estoque_unidades"] = pd.to_numeric(df["estoque_unidades"].astype("string").str.replace(" ", "",regex=False).str.replace("un","",regex=False))



print(df.info())
display(df.head(100))
```

```python
#Qual Fornecedor forneceu mais produtos

#Fatoramento
df["fatoramento_Produtos"] = df["preco"] * df["estoque_unidades"]
display(df)
#Top 5 fornecedores
lucro_fornecedor = df[["fornecedor","fatoramento_Produtos"]].groupby("fornecedor").sum()
#Top 5 Categorias
lucro_categoria = df[["categoria","fatoramento_Produtos"]].groupby("categoria").sum()
lucro_categoria = lucro_categoria.sort_values(by="fatoramento_Produtos", ascending=False)
display(lucro_fornecedor.head(5))
display(lucro_categoria.head(5))
```
